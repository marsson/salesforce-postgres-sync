```python
setup_logging()
salesforce = SalesforceAPI()

try:
    # Get metadata
    metadata = salesforce.fetch_object_metadata('Account')

    # Filter fields that are queryable
    queryable_fields = [
        field['name']
        for field in metadata['fields']
        if field.get('queryable', True)  # Only include queryable fields
        and not field['name'].endswith('__pc')  # Exclude computed fields
        and field['type'] != 'address'  # Exclude compound fields
    ]

    # Build SOQL query
    fields_string = ', '.join(queryable_fields)
    query = f"SELECT {fields_string} FROM Account LIMIT 10"

    print("Executing query:")
    print(query)

    # Execute the query
    results = salesforce.fetch_data(query)

    if results and 'records' in results:
        print(f"\nFound {len(results['records'])} records")
        for i, record in enumerate(results['records'], 1):
            record.pop('attributes', None)
            print(f"\nRecord {i}:")
            for field, value in record.items():
                if value is not None:  # Only show non-null values
                    print(f"{field}: {value}")

except Exception as e:
    print(f"Error: {e}")
```
